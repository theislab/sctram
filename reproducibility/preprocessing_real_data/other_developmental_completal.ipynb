```python
def combiner(adata_str, latent_str):
    adata_file_path = os.path.join(tardis.config.io_directories["processed"], f"dataset_complete_{adata_str}.h5ad")
    adata = ad.read_h5ad(adata_file_path)
    latent_scvi_path = f"/lustre/groups/ml01/workspace/kemal.inecik/tardis_data/_temporary/latent/{latent_str}_scvi_15_latent.h5ad"
    latent_scvi = ad.read_h5ad(latent_scvi_path)
    adata.obsm["scvi"] = latent_scvi.X.copy()
    adata.obs = latent_scvi.obs.copy()
    for k in latent_scvi.uns:
        adata.uns[k] = latent_scvi.uns[k].copy()
    for k in latent_scvi.obsm:
        adata.obsm[k] = latent_scvi.obsm[k].copy()
    for k in latent_scvi.obsp:
        adata.obsp[k] = latent_scvi.obsp[k].copy()
    working_dir = "/lustre/groups/ml01/workspace/kemal.inecik/sctram_data"
    file_name = f"{latent_str}_developmental_complete.h5ad"
    adata.write_h5ad(os.path.join(working_dir, file_name))
    display(adata)
```
