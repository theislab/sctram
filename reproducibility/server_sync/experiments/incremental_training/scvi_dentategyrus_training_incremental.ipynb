```python
# Manually download and put loom into dataset_dir
# go 'http://pklab.med.harvard.edu/velocyto/DentateGyrus/'
# click on safari and download. otherwise it just downloads an image of it.

# scvelo operations
adata = ad.read_loom(os.path.join(dataset_dir, "DentateGyrus.loom"))
adata.var_names_make_unique()
adata.obsm["X_tsne"] = np.column_stack([adata.obs["TSNE1"], adata.obs["TSNE2"]])
adata.obs["clusters"] = adata.obs["ClusterName"]
# cleanup
del adata.var, adata.layers, adata.obsm
gc.collect()
adata
```

```python
adata.layers["counts"] = adata.X.copy()
sc.pp.normalize_total(adata, target_sum=1e4) 
sc.pp.log1p(adata)                           
sc.pp.highly_variable_genes(
    adata,
    n_top_genes=5000,
    subset=True,          
    layer=None,            
    batch_key="SampleID"        
)
```

```python
step_per_epoch = 29  # gotten from the callback
print(step_per_epoch)

epochs = list(range(0, 400, 10))
epochs = list(np.arange(0, 10, 0.125)) + epochs
epochs = list(np.arange(10, 20, 0.5)) + epochs
# epochs = [round(i, 2) for i in epochs]
steps = sorted(set([int(i*step_per_epoch) for i in epochs]))

print(np.array(steps))
print(len(steps))
```

```python
cb_latent = LatentOnIndexCallback(
    indices=steps,
    capture_by="step",
    adata=adata,
    subset_idx=None,
    latent_dir=dataset_dir,
    latent_base_name="scvi_dentategyrus_incremental_with_callback"
)
logger_cb = EpochStepLogger()

model_params = dict(
    n_latent=24, 
    gene_likelihood = "nb",
)

train_params = dict(
    train_size=0.8,
    batch_size=512,
    limit_train_batches=1.00, 
    limit_val_batches=1.00,
    max_epochs=400,
)

dataset_params = dict(
    layer=None, 
    # labels_key="cell_type",
    # unlabeled_category="NA",
    batch_key="SampleID",
    categorical_covariate_keys=None,
)

scvi.model.SCVI.setup_anndata(adata, **dataset_params)
vae = scvi.model.SCVI(adata, **model_params)
vae.train(**train_params, callbacks=[cb_latent])  # callbacks=[logger_cb]

model_dir_path = os.path.join(dataset_dir, f"scvi_dentategyrus_incremental_with_callback")
vae.save(
    model_dir_path,
    overwrite=True,
)

pickle_paths_path = os.path.join(dataset_dir, f"scvi_dentategyrus_incremental_with_callback_paths.pkl")
with open(pickle_paths_path, "wb") as f:
    pickle.dump(cb_latent.latents, f, protocol=pickle.HIGHEST_PROTOCOL)

# as it is modified
adata.write_h5ad(os.path.join(dataset_dir, f"scvi_dentategyrus_incremental_with_callback_training_anndata.h5ad"))
    
# with open(pickle_paths_path, "rb") as f:
#     loaded_data: dict[str, str] = pickle.load(f)
```

```python
with allow_untrained(vae):
    latent = ad.AnnData(X=vae.get_latent_representation())
latent.obs = adata.obs.copy()
sc.pp.neighbors(latent)
sc.tl.umap(latent)

with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    sc.pl.umap(
        latent,
        color=["Age", "clusters"],
        ncols=4,
        frameon=False,
        legend_fontsize="xx-small",
        legend_loc="on data",
    )
```
