```python
adata = sc_suo_developmental_complete(dataset_dir=dataset_dir)
# display(adata)
adata = adata[adata.obs.LVL0 == "Haematopoeitic_lineage"]

def smart_subsample(adata, group_key="LVL3", max_cells_per_group=500, random_state=0):
    """
    Subsample overrepresented groups to 'max_cells_per_group' cells,
    retain all cells from underrepresented groups.
    """
    np.random.seed(random_state)
    groups = adata.obs[group_key]
    indices_to_keep = []

    for group in groups.unique():
        group_indices = adata.obs.index[groups == group].tolist()
        if len(group_indices) > max_cells_per_group:
            selected = np.random.choice(group_indices, max_cells_per_group, replace=False)
            indices_to_keep.extend(selected)
        else:
            indices_to_keep.extend(group_indices)

    return adata[indices_to_keep].copy()
# print()
# display(adata)
adata = smart_subsample(adata, max_cells_per_group=2000)
del adata.obsm, adata.uns
# print()
display(adata)
```

```python
step_per_epoch = 184  # gotten from the callback
print(step_per_epoch)

epochs = list(itertools.chain(range(0, 80, 1), range(80, 481, 40)))
epochs = list(np.arange(0, 20, 0.25)) + epochs
epochs = [round(i, 2) for i in epochs]
steps = sorted(set([int(i*step_per_epoch) for i in epochs]))

print(np.array(steps))
print(len(steps))
```

```python
cb_latent = LatentOnIndexCallback(
    indices=steps,
    capture_by="step",
    adata=adata,
    subset_idx=None,
    latent_dir=dataset_dir,
    latent_base_name="scvi_suo_incremental_with_callback"
)
logger_cb = EpochStepLogger()

model_params = dict(
    n_latent=24, 
    gene_likelihood = "nb",
)

train_params = dict(
    train_size=0.8,
    batch_size=512,
    limit_train_batches=1.00, 
    limit_val_batches=1.00,
    max_epochs=480,
)

dataset_params = dict(
    layer=None, 
    # labels_key="cell_type",
    # unlabeled_category="NA",
    batch_key="concatenated_integration_covariates",
    categorical_covariate_keys=None,
)

scvi.model.SCVI.setup_anndata(adata, **dataset_params)
vae = scvi.model.SCVI(adata, **model_params)
vae.train(**train_params, callbacks=[cb_latent])

model_dir_path = os.path.join(dataset_dir, f"scvi_suo_incremental_with_callback")
vae.save(
    model_dir_path,
    overwrite=True,
)

pickle_paths_path = os.path.join(dataset_dir, f"scvi_suo_incremental_with_callback_paths.pkl")
with open(pickle_paths_path, "wb") as f:
    pickle.dump(cb_latent.latents, f, protocol=pickle.HIGHEST_PROTOCOL)

with open(pickle_paths_path, "rb") as f:
    loaded_data: dict[str, str] = pickle.load(f)
```

```python
input_trajectories_path = os.path.join(dataset_dir, f"adata_suo_input_haematopoeitic_lineage.pkl")
with open(input_trajectories_path, "rb") as _file:
    lit = pickle.load(_file)
    cell_types = set(list(lit.get_trajectory("early_b_cells", False).nodes()))  # erythroid_megakaryocyte
np.array(cell_types)
```

```python
latent = ad.AnnData(X=vae.get_latent_representation())
latent.obs = adata.obs.copy()
sc.pp.neighbors(latent)
sc.tl.umap(latent)
latent_subset = latent[latent.obs.cell_type.isin(cell_types)]

with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    sc.pl.umap(
        latent,
        color=["age", "LVL0", "LVL1", "LVL2"],
        ncols=4,
        frameon=False,
        legend_fontsize="xx-small",
    )
with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    sc.pl.umap(
        latent_subset,
        color=["age", "cell_type"],
        ncols=4,
        frameon=False,
        legend_fontsize="xx-small",
    )
```
